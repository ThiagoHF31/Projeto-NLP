```python
import importlib.util
import subprocess
import sys
import random
from pathlib import Path

# Garante PyMuPDF instalado antes do import
if importlib.util.find_spec("fitz") is None:
    print("Instalando PyMuPDF...")
    subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", "pymupdf"])

import fitz  # Importa o PyMuPDF

# Caminho ajustado para ser relativo à estrutura do projeto
pasta_pdfs = Path("../dados/pdfs")

# Pega todos os PDFs e escolhe 3 aleatórios para não viciar o teste
todos_pdfs = list(pasta_pdfs.glob("*.pdf"))

if not todos_pdfs:
    print("❌ Nenhum PDF encontrado. Verifique se o caminho '../dados/pdfs' está correto.")
else:
    pdfs_teste = random.sample(todos_pdfs, min(3, len(todos_pdfs)))

    print("🔬 INICIANDO O TESTE DE RAIO-X NOS PDFs...\n" + "="*60)

    for i, caminho_pdf in enumerate(pdfs_teste, 1):
        print(f"\n📄 ARQUIVO {i}: {caminho_pdf.name}")
        
        try:
            # Abre o documento
            documento = fitz.open(caminho_pdf)
            
            # Pega o número total de páginas e seleciona a primeira (índice 0)
            total_paginas = len(documento)
            pagina = documento[0]
            
            # Extrai o texto preservando ao máximo a ordem visual
            texto_extraido = pagina.get_text("text")
            
            # Puxa informações sobre imagens na página
            imagens = pagina.get_images()
            
            print(f"📊 Total de páginas no arquivo: {total_paginas}")
            print(f"🖼️ Imagens detectadas nesta página: {len(imagens)}")
            print("-" * 60)
            
            # Diagnóstico de OCR
            if len(texto_extraido.strip()) < 50 and len(imagens) > 0:
                print("⚠️ DIAGNÓSTICO: ALERTA DE OCR!")
                print("Quase não há texto selecionável, mas há imagens. Este PDF provavelmente é um documento escaneado.")
            elif len(texto_extraido.strip()) == 0:
                print("⚠️ DIAGNÓSTICO: PÁGINA EM BRANCO ou formato desconhecido.")
            else:
                print("✅ DIAGNÓSTICO: TEXTO ENCONTRADO! Aqui está a prévia (primeiros 800 caracteres):\n")
                # Mostra apenas os primeiros 800 caracteres para não inundar o terminal
                print(texto_extraido[:800] + "\n\n[... CONTINUA ...]")
                
            documento.close()
            
        except Exception as e:
            print(f"❌ ERRO ao tentar ler o arquivo: {e}")
            
        print("="*60)
```
